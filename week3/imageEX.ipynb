```python
import matplotlib.pyplot as plt
from sklearn import datasets, metrics
from sklearn.model_selection import train_test_split

digits = datasets.load_digits()
```

```python
plt.imshow(
    digits.images[0],
    cmap=plt.cm.gray_r,
    interpolation='nearest'
)
```

```python
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.2
)
```

```python
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)

scores = metrics.accuracy_score(y_test, y_pred)
print(scores)
```

```python
plt.imshow(
    X_test[10].reshape(8, 8),
    cmap=plt.cm.gray_r,
    interpolation='nearest'
)

y_pred = knn.predict([X_test[10]])
print(y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier

# 1. Decision Tree (결정 트리)
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

dt_pred = dt.predict(X_test)
scores = metrics.accuracy_score(y_test, dt_pred)
print(scores)

plt.imshow(
    X_test[10].reshape(8, 8),
    cmap=plt.cm.gray_r,
    interpolation='nearest'
)

dt_pred = dt.predict([X_test[10]])
print(dt_pred)
```

```python
from sklearn.linear_model import LogisticRegression
# 2. Logistic Regression (로지스틱 회귀)
lr = LogisticRegression(max_iter=5000, random_state=42)
lr.fit(X_train, y_train)

lr_pred = lr.predict(X_test)
scores = metrics.accuracy_score(y_test, lr_pred)
print(scores)

plt.imshow(
    X_test[10].reshape(8, 8),
    cmap=plt.cm.gray_r,
    interpolation='nearest'
)

lr_pred = lr.predict([X_test[10]])
print(lr_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
# 3. Random Forest (랜덤 포레스트)
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)
scores=metrics.accuracy_score(y_test, rf_pred)
print(scores)

plt.imshow(
    X_test[10].reshape(8, 8),
    cmap=plt.cm.gray_r,
    interpolation='nearest'
)

rf_pred = rf.predict([X_test[10]])
print(rf_pred)
```
